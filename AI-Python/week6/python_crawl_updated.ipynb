## 1.HTML гэж юу вэ?

Вэб хуудсыг HTML хэлний тусламжтай хийдэг. HTML хэл нь script хэл бөгөөд ямар нэг хөрвүүлэлт шаарддаггүй, дурын текст боловсруулах програм дээр бичиж болдог. Харин тухайн файлаа *.html юмуу *.htm форматтай хадгалах хэрэгтэй. Дараа нь уг файлаа Вэб браузер програм дээр нээж ажиллуулах хэрэгтэй.
HTML хэлний командуудыг TAG гэж нэрлэх ба TAG нь эхлэл ба төгсгөлөөс тогтоно. Эхлэл нь <…> хашилтан дотор, төгсгөл нь хашилтан дотор тус тус байрладаг. Гүйцэтгэх үйлдлээ эхлэл ба төгсгөлийн хооронд бичнэ. Жишээлбэл, хуудсанд гарчиг оруулахын тулд TITLE тагийг ашиглах бөгөөд дараах байдлаар бичнэ.
HTML тагуудыг хэдэн хэдэн ангилалд оруулан авч үзэж болно. Үүнд:
1. Html Basics–Вэб хуудасны үндсэн тагууд,
2. Html Text–Текстэн мэдээлэлтэй ажиллах тагууд,
3. Html Lists–Жагсаалт үүсгэх тагууд,
4. Html Images–Зурагтай ажиллах тагууд,
5. Html Links–Холбоос тагууд,
6. Html Backgrounds–Вэб хуудасны дэвсгэрийг тохируулах тагууд,
7. Html Tables–Хүснэгттэй ажиллах тагууд,
8. Html Frames–Frame-тэй ажиллах тагууд,
9. Html Forms–Form, анкеттай ажиллах тагууд,
10. Html Metatags–Хайлтын систем болон браузертай ажиллах тагууд,
11. Html Hexcolors-Өнгөтэй холбоотой тагууд.

Үндсэн HTML хэлний бүтэц нь толгойн хэсэг ба их биеийн хэсгээс тогтох ба HTML, HEAD, BODY гэсэн үндсэн 3 тагийг агуулах бөгөөд дараах хэлбэртэй байна. Үүнд:



. . .
. . . Толгойн хэсэг
. . .

. . .
. . . Их биеийн хэсэг
. . .




Энд:
- … таг нь Вэб хуудасны эхлэл, төгсгөлийг,
- … таг нь толгойн хэсэг буюу Вэб хуудасны гарчиг гэх мэт элементүүдийг,
- … таг нь их биеийн хэсэг буюу HTML хэл дээр бичигдэж байгаа ихэнх эх кодыг агуулна.
Тухайн тохиолдолд Вэб хуудсанд ашиглагдсан Javascript, ActiveX Script гэх мэт бусад Script хэлний код HEAD таг дотор тодорхойлогдож болно.
HTML тагууд нь мөн параметр (option)-ууд агуулж болох ба энэ нь хуудсыг нэмэлт боломжуудаар хангаж өгдөг.

# Table scraping

```python
table = soup.find("table")
table
```

```python
columns = table.find("tr").find_all("th")
columns
columns_text = []
for col in columns:
    columns_text.append(col.text)
columns_text
```

```python
table_rows = table.find_all("tr")[1:]
table_dictionary = []
for table_row in table_rows:
    row_dictionary = {}
    row_dictionary[columns_text[0]] = table_row.find_all("td")[0].text
    row_dictionary[columns_text[1]] = table_row.find_all("td")[1].text
    row_dictionary[columns_text[2]] = table_row.find_all("td")[2].text
    table_dictionary.append(row_dictionary)
table_dictionary
```

```python
import pandas as pd

table_df = pd.DataFrame.from_dict(table_dictionary)
table_df
```

```python
import requests
from bs4 import BeautifulSoup
import time

home_url = "https://www.unegui.mn/l-hdlh/l-hdlh-zarna/"
names_list = []
prices_list = []
url_list = []
whens_list = []
where_list = []
for i in range(1,6):
    print(i)
    page_url = home_url + "?page=" + str(i)
    reqs = requests.get(page_url)
    soup = BeautifulSoup(reqs.content, 'html.parser')
    time.sleep(3)

    urls = soup.find_all("a", class_="mask")
    for url in urls:
        if url["href"] == None:
            url_list.append("")
        else:
            url_list.append(home_url[:-1]+url["href"])
    
    names = soup.find_all("a", class_="advert__content-title")
    for name in names:
        if name.text.strip() == None:
            names_list.append("")
        else:
            names_list.append(name.text.strip())

    prices = soup.find_all("a", class_="advert__content-price _not-title")
    for price in prices:
        if price.text.strip() == None:
            prices_list.append("")
        else:
            prices_list.append(price.text.strip())

    whens = soup.find_all("div", class_="advert__content-date")
    for when in whens:
        if when.text.strip() == None:
            whens_list.append("")
        else:
            whens_list.append(when.text.strip())

    wheres = soup.find_all("div", class_="advert__content-place")
    for where in wheres:
        if where.text.strip() == None:
            where_list.append("")
        else:
            where_list.append(where.text.strip())
```

```python
df = pd.DataFrame()
df["url"] = url_list
df["name"] = names_list
df["price"] = prices_list
df["hezee"] = where_list
df["haana"] = whens_list
df.to_excel("/Users/ddam-m0101/Downloads/sample_unegui_2.xlsx", index = False)
```

```python
from bs4 import BeautifulSoup
import time

output_lis = []
for ind, url in enumerate(data["url"][:50]):
    print(ind)
    reqs = requests.get(url)
    soup = BeautifulSoup(reqs.content, 'html.parser')
    table = soup.find(class_ = "chars-column")
    table_list = table.find_all("li")
    table_dic = {}
    for table_element in table_list:
        table_key = table_element.find(class_ ="key-chars").text.strip()
        table_value = table_element.find(class_ ="value-chars").text.strip()
        table_dic[table_key]=table_value
    time.sleep(2)
    output_lis.append(table_dic)
    
```

```python
df_dict = pd.DataFrame.from_dict(output_lis)
df_dict
```
