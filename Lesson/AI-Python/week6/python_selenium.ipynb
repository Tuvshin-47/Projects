```python
from selenium.webdriver.common.keys import Keys

driver = webdriver.Chrome()

url = 'https://www.unegui.mn/l-hdlh/l-hdlh-zarna/'
driver.get(url)
name_list = []
url_list = []
for i in range(1,21):
    print(i)
    try:
        name_xpath = "/html/body/div[2]/div[3]/section/div[2]/div[1]/div[2]/div[2]/div["+str(i)+"]/div[2]/div[2]/a"
        element_name = driver.find_element(By.XPATH, name_xpath)
    except:
        name_xpath = "/html/body/div[2]/div[3]/section/div[2]/div[1]/div[2]/div[2]/div["+str(i)+"]/div[2]/div/a"
        element_name = driver.find_element(By.XPATH, name_xpath)
    name = element_name.text
    url = element_name.get_attribute('href')
    name_list.append(name)
    url_list.append(url)
```
